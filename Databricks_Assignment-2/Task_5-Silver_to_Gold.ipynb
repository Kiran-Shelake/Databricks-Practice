```python
from pyspark.sql.functions import sum, max, round, desc, year, month

#--> Sales by state or region
df_sales_by_state = spark.table("main.ecommerce.fact_sales"). groupBy("customer_state").agg(sum("revenue").alias("total_sales"))
display(df_sales_by_state)

df_sales_by_state.write.format("delta").mode("overwrite").saveAsTable("main.ecommerce.sales_by_states")

df_sales_by_state.write.mode("overwrite").save("/Volumes/main/ecommerce/lakehouse_vol/Gold/sales_by_state/")
```

```python
# --> Top products by revenue
df_top_products_by_revenue = spark.table("main.ecommerce.fact_sales").groupBy("product_category_name").agg(round(sum("revenue"),2).alias("Total_revenue")).orderBy(desc("Total_revenue"))

display(df_top_products_by_revenue)

df_top_products_by_revenue.write.format("delta").mode("overwrite").saveAsTable("main.ecommerce.top_products_by_revenue")
df_top_products_by_revenue.write.mode("overwrite").save("/Volumes/main/ecommerce/lakehouse_vol/Gold/top_products_by_revenue/")
```

```python
# --> Daily or monthly sales trends
#On Daily basis
df_daily_sale = spark.table("main.ecommerce.fact_sales").groupBy("order_date").agg(round(sum("revenue"),2).alias("daily_revenue")).orderBy("order_date")
display(df_daily_sale)

df_daily_sale.write.format("delta").mode("overwrite").saveAsTable("main.ecommerce.daily_sale")

df_daily_sale.write.mode("overwrite").save("/Volumes/main/ecommerce/lakehouse_vol/Gold/daily_sale/")

#On Montly basis
df_monthly_sale = spark.table("main.ecommerce.fact_sales").groupBy(year("order_date").alias("year"), month("order_date").alias("month")).agg(round(sum("revenue"),2)).orderBy("year","month")

display(df_monthly_sale)

df_daily_sale.write.format("delta").mode("overwrite").saveAsTable("main.ecommerce.monthly_sale")

df_daily_sale.write.mode("overwrite").save("/Volumes/main/ecommerce/lakehouse_vol/Gold/monthly_sale/")
```
