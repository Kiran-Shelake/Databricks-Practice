```python
Source_Path = "/Volumes/main/ecommerce/lakehouse_vol/Raw_Data/"
Target_Path = "/Volumes/main/ecommerce/lakehouse_vol/Bronze/"

for file in dbutils.fs.ls(Source_Path):
#Read operation --> Raw   
    df = spark.read.csv(file.path, header = True, inferSchema = True)

    Split_Name = file.name.split('_')
    Folder_Name = '_'.join(Split_Name[1:-1])
    #print(Folder_Name)
    Desination_Path = Target_Path + Folder_Name
    #File_name = file.name.replace('.csv', '')

#Write Operation --> Bronze
    df.write.mode("overwrite").parquet(str(Desination_Path))
  
```
