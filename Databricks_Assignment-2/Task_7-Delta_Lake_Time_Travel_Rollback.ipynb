```python
%sql
-- #View Delta table history

DESCRIBE HISTORY main.ecommerce.fact_sales;
```

```python
%sql
-- #Query previous versions of a table
SELECT * from main.ecommerce.fact_sales VERSION AS OF 1;
```

```python
%sql
-- #Simulate an incorrect update or delete
UPDATE main.ecommerce.fact_sales SET revenue = 0; --> will set column value to 0. Incorrect update  
DELETE FROM main.ecommerce.fact_sales; --> will delete all rows fof the table. Incorrect delete
```

```python
%sql
-- #Restore the table using:
--> Version OR
--DESCRIBE HISTORY main.ecommerce.fact_sales
RESTORE TABLE main.ecommerce.fact_sales TO VERSION AS OF 1;
--> Timestamp
RESTORE TABLE main.ecommerce.fact_sales TO TIMESTAMP AS OF '2025-12-31 13:55:54';
```

```python
%sql
-- #Validate successful rollback
-- DO row count check
SELECT count(*) FROM main.ecommerce.fact_sales;

-- Do Revenue validation
SELECT MIN(revenue), MAX(revenue) from main.ecommerce.fact_sales;
```
